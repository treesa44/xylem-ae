```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import butter, lfilter, spectrogram
from scipy.io import wavfile

FS = 250000            # ESP32-S3 ADC continuous-DMA sample rate
CLICK_FREQ = 113000     # piezo disc resonance
DURATION = 2.0
t = np.linspace(0, DURATION, int(FS * DURATION), endpoint=False)

print(f"Sampling Rate: {FS/1000:.1f} kHz | Nyquist: {FS/2/1000:.1f} kHz | Click: {CLICK_FREQ/1000:.1f} kHz")

def generate_click(t, center_freq=CLICK_FREQ, burst_duration=0.002, amplitude=1.0):
    click = np.zeros_like(t)
    start = np.random.uniform(0.2, DURATION - 0.2)
    idx = np.where((t >= start) & (t < start + burst_duration))
    if len(idx[0]) > 0:
        envelope = np.hanning(len(idx[0]))
        click[idx] = amplitude * envelope * np.sin(2 * np.pi * center_freq * t[idx])
    return click, start

def generate_wind_noise(t, amplitude=0.8):
    # fresh random draw each call -> independent noise realizations
    white = np.random.randn(len(t))
    b, a = butter(4, 1000 / (FS / 2), btype='low')
    colored = lfilter(b, a, white)
    colored = colored / np.max(np.abs(colored))
    return amplitude * colored

def generate_insect_noise(t, amplitude=0.3, num_chirps=30):
    noise = np.zeros_like(t)
    for _ in range(num_chirps):
        start = np.random.uniform(0.1, DURATION - 0.1)
        dur = np.random.uniform(0.005, 0.03)
        freq = np.random.uniform(2000, 10000)
        idx = np.where((t >= start) & (t < start + dur))
        if len(idx[0]) > 0:
            envelope = np.hanning(len(idx[0]))
            noise[idx] += amplitude * envelope * np.sin(2 * np.pi * freq * t[idx])
    return noise

# --- Generate signals ---
print("\nGenerating synthetic data...")
click_signal = np.zeros_like(t)
click_times_ground = []
num_clicks = np.random.randint(5, 12)
for _ in range(num_clicks):
    click, start_time = generate_click(t)
    click_signal += click
    click_times_ground.append(start_time)

# Plant-side (stem) noise components
wind_plant = generate_wind_noise(t, amplitude=0.8)
insect_plant = generate_insect_noise(t, amplitude=0.3)

# Reference-side (wind-facing mic) noise: correlated but NOT identical --
# a second, independent draw mixed in simulates two physically separate sensors
wind_ref = generate_wind_noise(t, amplitude=0.8)
insect_ref = generate_insect_noise(t, amplitude=0.3)

plant_mic = click_signal + wind_plant + insect_plant
ref_mic = 0.85 * wind_plant + 0.3 * wind_ref + 0.85 * insect_plant + 0.3 * insect_ref

# --- Normalize BEFORE int16 conversion to prevent clipping/wraparound ---
plant_mic = plant_mic / np.max(np.abs(plant_mic))
ref_mic = ref_mic / np.max(np.abs(ref_mic))

print(f"Generated {num_clicks} clicks at {CLICK_FREQ/1000:.1f} kHz")
print(f"Click times: {[round(c,3) for c in click_times_ground]}")

wavfile.write('plant_mic.wav', FS, (plant_mic * 32767).astype(np.int16))
wavfile.write('ref_mic.wav', FS, (ref_mic * 32767).astype(np.int16))
np.save('click_times_ground_truth.npy', np.array(click_times_ground))
print("Saved: plant_mic.wav, ref_mic.wav, click_times_ground_truth.npy")

# --- Plot ---
fig, axes = plt.subplots(2, 2, figsize=(14, 7))
show = int(0.05 * FS)
axes[0,0].plot(t[:show]*1000, plant_mic[:show]); axes[0,0].set_title("Plant mic (clicks+noise)")
axes[0,1].plot(t[:show]*1000, ref_mic[:show]); axes[0,1].set_title("Reference mic (noise only)")
f, ts, Sxx = spectrogram(plant_mic, fs=FS, nperseg=256, noverlap=200)
axes[1,0].pcolormesh(ts, f/1000, 10*np.log10(Sxx+1e-10), shading='gouraud'); axes[1,0].set_ylim([0,125])
axes[1,0].set_title("Plant mic spectrogram"); axes[1,0].set_ylabel("kHz")
f, ts, Sxx = spectrogram(ref_mic, fs=FS, nperseg=256, noverlap=200)
axes[1,1].pcolormesh(ts, f/1000, 10*np.log10(Sxx+1e-10), shading='gouraud'); axes[1,1].set_ylim([0,125])
axes[1,1].set_title("Reference mic spectrogram"); axes[1,1].set_ylabel("kHz")
plt.tight_layout()
plt.show()
```
